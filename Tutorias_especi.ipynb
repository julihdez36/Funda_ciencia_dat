```python
# Para reflexionar: cuántas veces aparece el 2 en la lista

suma = 0
for i in lista1:
    suma += [str(i) for i in lista1].count(2)
suma


lista_20_60 = list(range(20, 61, 2))
frecuencia_numeros_repetidos = {}
for numero in lista_20_60:
    if numero in frecuencia_numeros_repetidos:
        frecuencia_numeros_repetidos[numero] += 1
    else:
        frecuencia_numeros_repetidos[numero] = 1
print("Frecuencia de números: " + str(frecuencia_numeros_repetidos))
```
